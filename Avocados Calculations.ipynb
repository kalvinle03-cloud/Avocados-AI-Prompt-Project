# Price Trends (Conventional)


```python
import pandas as pd
```

```python
df_2025 = pd.read_csv("2025Data.csv")
df_2026 = pd.read_csv("2026Data.csv")
```

```python
df_2025['Year'] = 2025
df_2026['Year'] = 2026
df_combined = pd.concat([df_2025, df_2026], ignore_index=True)
```

```python
df_filtered = df_combined[(df_combined['Geography'] == 'Total U.S.') & 
(df_combined['Type'] == 'Conventional')]
```

```python
asp_2025 = df_filtered[df_filtered['Year'] == 2025]['ASP Current Year'].mean()
asp_2026 = df_filtered[df_filtered['Year'] == 2026]['ASP Current Year'].mean()
```

```python
pct_change = ((asp_2026 - asp_2025) / asp_2025) * 100
```

```python
print(f"2025 Average Selling Price: {asp_2025}")
print(f"2026 Average Selling Price: {asp_2026}")
print("-" * 40)
print("Calculation for Percentage Decline:")
print(f"(({asp_2026} - {asp_2025}) / {asp_2025}) * 100")
print(f"Result: {pct_change}%")
```

# Volume and Demand Calculation (Conventional)


```python
df_2025 = pd.read_csv("2025Data.csv")
df_2026 = pd.read_csv("2026Data.csv")
```

```python
df_2025['Year'] = 2025
df_2026['Year'] = 2026
df_combined = pd.concat([df_2025, df_2026], ignore_index=True)
```

```python
df_filtered = df_combined[(df_combined['Geography'] == 'Total U.S.') & 
(df_combined['Type'] == 'Conventional')]
```

```python
total_vol_2025 = df_filtered[df_filtered['Year'] == 2025]['Total Bulk and Bags'].sum()
total_vol_2026 = df_filtered[df_filtered['Year'] == 2026]['Total Bulk and Bags'].sum()
```

```python
weeks_2025 = 52
weeks_2026 = 31
weekly_avg_2025 = total_vol_2025 / weeks_2025
weekly_avg_2026 = total_vol_2026 / weeks_2026
```

```python
pct_change_vol = ((weekly_avg_2026 - weekly_avg_2025) / weekly_avg_2025) * 100
```

```python
print(f"2025 Total Volume: {total_vol_2025:,.0f} (across {weeks_2025} weeks)")
print(f"2026 YTD Volume: {total_vol_2026:,.0f} (across {weeks_2026} weeks)")
print("-" * 40)
print(f"2025 Weekly Avg: {weekly_avg_2025:,.0f}")
print(f"2026 Weekly Avg: {weekly_avg_2026:,.0f}")
print("-" * 40)
print("Calculation for Volume Percentage Increase:")
print(f"(({weekly_avg_2026:.0f} - {weekly_avg_2025:.0f}) / {weekly_avg_2025:.0f}) * 100")
print(f"Result: +{pct_change_vol:.1f}%")
```
